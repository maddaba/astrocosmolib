Qui si lavora con file FITS usando la classe FitsManager, creata per rendere più semplici le operazioni rispetto all’uso diretto di astropy. Viene caricato il file "HorseHead.fits", da cui si estraggono e visualizzano i dati dell’immagine.

```python
import sys  
import os
import matplotlib.pyplot as plt

sys.path.append("../../") # Aggiunge al path la cartella due livelli sopra
from pyACC.fits_wrapper import FitsManager # Importa la classe FitsManager

input_file= "/home/maddaba/astrocosmolib/Esercizi/fits/HorseHead.fits"
```

```python
horse_head_fits=FitsManager(input_file) # Crea un'istanza di FitsManager con il file FITS
```

```python
horse_head_fits.input_file # Specifica il percorso completo al file FITS da aprire
```

```python
horse_head_fits.hdulist 
```

```python
horse_head_fits.get_hdu_count() # Restituisce il numero di HDU nel file FITS
```

```python
data = horse_head_fits.get_data(0) # Estrae i dati dell’HDU all’indice 0 
plt.imshow(data, cmap='gray')
plt.colorbar()
plt.title("Horse Head Nebula")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
