Questo esercizio mostra come aprire un file FITS, esplorarne la struttura e i metadati, ed estrarre i dati contenuti. Utilizzando il file "HorseHead.fits", si visualizza l'immagine astronomica in esso contenuta tramite la libreria matplotlib.

```python
import numpy as numpy # Importa il modulo NumPy per gestire array numerici  
import matplotlib.pyplot as plt # Importa il modulo Matplotlib per creare grafici
from astropy.io import fits # Importa il modulo per leggere file FITS
```

```python
fits_file="HorseHead.fits"
fits_hdu = fits.open(fits_file)  # Apre il file FITS e lo carica in una lista di HDU
print(fits_hdu)
```

```python
fits_hdu.info() # Stampa le informazioni sugli HDU presenti nel file FITS
```

```python
fits_hdu[0].header # Stampa l'header del primo HDU per visualizzare le informazioni di metadati
```

```python
data=fits_hdu[0].data  # Estrae i dati dal primo HDU, che contiene l'immagine
cb = plt.imshow(data,cmap='viridis')  # Visualizza i dati come un'immagine con la mappa dei colori 'viridis'
plt.title("Horse Head Nebula")  
plt.colorbar(cb) # Aggiunge una barra dei colori per indicare la scala dei valori
plt.xlabel("X")
plt.xlabel("Y")
```
