Questo esercizio simula due insiemi di punti distribuiti casualmente in un cubo (tipo universo finto) e calcola tra di loro la funzione di correlazione a due punti xi usando la libreria TwoPointIsotropic. L’obiettivo è capire se ci sono strutture o clustering anche in una distribuzione puramente casuale.

```python
import sys
import numpy as np
import matplotlib.pyplot as plt
sys.path.append("../../")


from pyACC.clustering.measure.two_point import TwoPointIsotropic  
```

```python
L = 1000 # Dimensioni del cubo
n = 100000 # Numero di punti
x1, y1, z1 = np.random.rand(3, n) * L   # Primo set di punti distribuiti uniformemente nel cubo
x2, y2, z2 = np.random.rand(3, n) * L   # Secondo set
```

```python
twopcf = TwoPointIsotropic()  # Creo un oggetto della classe TwoPointIsotropic per il calcolo di xi(r)
r_min = 0
r_max = 50
delta_r = 1 # Ampiezza dei bin

# Calcolo la funzione di correlazione
twopcf.compute(r_min, r_max, delta_r, x1, y1, z1, x2, y2, z2, periodic=True, boxsize=L)

plt.plot(twopcf.r_centers, twopcf.xi, label="Two-point correlation function")
plt.xlabel(r"$r$ [Mpc/h]")
plt.ylabel(r"$\xi(r)$")
plt.title("Two-point correlation function")
plt.show()
```

Come ci si aspetta da una distribuzione completamente casuale, xi(r) è circa zero per quasi tutte le distanze: non c'è clustering.
